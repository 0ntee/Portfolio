# Кинематографический анализ MovieLens

Кино — на удивление молодое искусство: чуть больше века против тысячелетий устной традиции. По сути это прямое продолжение одного из древнейших человеческих занятий — рассказывать истории у костра, вместо которого теперь экран.

Новизна нашего времени не в самих историях, а в масштабе наблюдения за реакцией на них. Раньше о желаниях и мечтах общества судили по обрывочным свидетельствам — письмам, дневникам, кассовым сборам горстки театров. Сегодня миллионы людей сами, добровольно и в деталях, документируют свою реакцию на истории: ставят оценку, пишут тег, выбирают, что пересмотреть. MovieLens — один из таких архивов коллективной реакции.

В этом отчёте мы используем его, чтобы увидеть отражение — не только индустрии, но и нас самих: чего люди ищут в историях, от чего устают, чего с любопытством ждут и на что готовы тратить деньги. Разберём это в четырёх актах: эпохи и жанры, теги и ассоциации, оценки зрителей, экономика индустрии.

---

```python
import movielens_analysis

%timeit movielens_analysis
```

```python
DATA_DIR = "datasets/"

movies = movielens_analysis.Movies(DATA_DIR + "movies.csv")
tags = movielens_analysis.Tags(DATA_DIR + "tags.csv")
ratings = movielens_analysis.Ratings(DATA_DIR + "ratings.csv", DATA_DIR + "movies.csv")
links = movielens_analysis.Links(DATA_DIR + "links.csv", DATA_DIR + "movies.csv")
dvis = movielens_analysis.DataViz()

%timeit movies, tags, ratings, links, dvis
```

# Акт I: Эпохи и жанры

Жанр — это ответ индустрии на вопрос «что тебе сейчас нужно почувствовать?». Прежде чем смотреть на оценки и деньги, разберёмся с этим фундаментом: когда и что именно снимали, и что выбор жанров говорит о коллективных потребностях аудитории.

### ❓ Вопрос 1: В какие годы мир увидел больше всего премьер?

```python
dvis.bar_chart(
    movies.dist_by_release(),
    title="Movie Distribution by year",
    max_bar_width=30,
)

%timeit movies.dist_by_release()
```

> Никаких девяностых: заветный пик — **2002 год, 311 премьер**, почти вровень со всей серединой нулевых. Хотя расцвет видеопроката принято относить к 90-м, реальный вал релизов пришёлся на цифровую эпоху: удешевление съёмки и монтажа открыло дорогу независимому кино.

Если посмотреть на общую картину, окажется, что подавляющему большинству от историй нужно всего два состояния: порадоваться и посмеяться — или прочувствовать и порефлексировать.

### ❓ Вопрос 2: Какие жанры были популярны на протяжении всей истории кино?

```python
dvis.bar_chart(
    movies.dist_by_genres(),
    title="Genres by Frequency",
    max_bar_width=30,
)

%timeit movies.dist_by_genres()
```

> Итог предсказуем и в то же время красноречив: драма и комедия вместе — это **почти 37% всего каталога**. Это не значит, что зрители не любят фантастику или хоррор — просто два этих жанра оказались самыми дешёвыми в производстве и самыми универсальными по спросу, поэтому индустрия штампует их из десятилетия в десятилетие.

Иногда привычных состояний — посмеяться или порефлексировать — недостаточно. Временами хочется чего-то необузданного, что не укладывается ни в одну привычную категорию.

### ❓ Вопрос 3: Какие фильмы объединили в себе рекордное число жанров сразу?

```python
dvis.bar_chart(
    movies.most_genres(5),
    title="Top 5 Movies by Genre Count",
    max_bar_width=25,
)

%timeit movies.most_genres(5)
```

> *Rubber* (2010, в русском прокате — «Шина») — фильм про шину-убийцу, обретшую сознание, — единственный в выборке собрал сразу 10 жанровых ярлыков. Формально это можно списать на то, что классификаторам не хватило существующих категорий. Но по сути это и есть выход за рамки привычного: усталость от знакомых форм иногда ищет разрядку в намеренно абсурдном и неклассифицируемом.

Если жанровый хаос — редкое исключение, то большинство фильмов, наоборот, комбинируют жанры по проверенным формулам.

### 💡 Бонусный вопрос 4: Какие жанровые группы стали золотым стандартом киноиндустрии?

```python
dvis.bar_chart(
    movies.genre_combinations(10),
    title="Top 10 Genre Combinations",
    max_bar_width=30,
)

%timeit movies.genre_combinations(10)
```

> Оказывается, «чистые» жанры без примесей популярнее любых гибридов: фильмы, помеченные просто как `Drama` или просто как `Comedy`, обходят даже привычную формулу romcom `Comedy|Romance`. Похоже, публика ценит фильм, который точно знает, чем он хочет быть.

# Акт II: Теги и ассоциации

Жанры — это официальная классификация, придуманная индустрией. Теги — куда менее формальный слой: то первое слово, которое зритель сам, без всякой инструкции, счёл нужным оставить после титров. Жанры показывают, что индустрия производит, теги говорят, что из этого действительно откладывается у людей в голове.

### ❓ Вопрос 5: Какие теги всплывают чаще всего?

```python
dvis.bar_chart(
    tags.most_popular(10),
    title="Top 10 Most Popular Tags",
    max_bar_width=30,
)

%timeit tags.most_popular(10)
```

> Самый популярный тег — вообще не про кино: это служебная пометка `In Netflix queue` (131 раз) — люди тегировали не эмоцию, а напоминалку самим себе. Зато сразу за ней идут куда более человечные слова — `atmospheric` и `thought-provoking`. Похоже, зрителям MovieLens важнее атмосфера и пища для ума, чем спецэффекты.

Иногда одного слова не хватает — и рождается мини-рецензия прямо в поле для тега.

### ❓ Вопрос 6: Какие теги имеют наибольшее число слов и символов?

```python
dvis.bar_chart(
    tags.most_words(10),
    title="Top 10 Tags by Word Count",
    max_bar_width=30,
)

%timeit tags.most_words(10)
```

```python
display(tags.longest(10))

%timeit tags.longest(10)
```

> Рекордсмен по многословности — признание длиной в целое предложение: *«Something for everyone in this one... saw it without and plan on seeing it with kids!»* — шестнадцать слов чистой родительской рефлексии вместо тега.

А теперь взглянем на пересечение: какие теги одновременно состоят из множества слов и имеют рекордную длину в символах?

```python
display(tags.most_words_and_longest(5))

%timeit tags.most_words_and_longest(5)
```

Некоторые теги — это прямое указание на культурный ориентир: студию, франшизу, целый пласт воображения. Показательный пример ниже — поиск по слову «robots»: он поднимает истории про андроидов и механических существ, то самое любопытство к будущему, что мы уже видели в жанровой статистике (Sci-Fi, Fantasy).

### ❓ Вопрос 7: Как зрители описывают знаковые вселенные и студии (например, 'pixar')?

```python
display(tags.tags_with("pixar"))

%timeit tags.tags_with("pixar")
```

```python
display(tags.tags_with("robots"))

%timeit tags.tags_with("robots")
```

> Оба найденных тега — `ROBOTS AND ANDROIDS` и `robots` — подтверждают то, что уже видно в жанровой статистике: истории о будущем, ИИ и механических существах остаются устойчивым источником сюжетов на протяжении всей истории каталога, а не проходящей модой одного десятилетия.

Одни зрители ограничиваются словом-меткой, другие превращают тег в маленькое высказывание. Кто из них самый изобретательный?

### 💡 Бонусный вопрос 8: Кто из пользователей обладает самым богатым и неординарным вокабуляром?

```python
dvis.bar_chart(
    tags.user_vocabulary_richness(10),
    title="Top 10 Users by Richness",
    max_bar_width=30,
)

%timeit tags.user_vocabulary_richness(10)
```

> Здесь статистическая ловушка: у десятков пользователей «идеальная» richness = 1.0. На деле это означает не богатство словаря, а то, что человек поставил всего один тег за всю историю — при выборке из одного элемента показатель разнообразия математически не может быть иным. Хороший повод помнить, что красивая метрика не всегда означает то, что кажется на первый взгляд.

# Акт III: Взгляд сквозь оценки

Жанры и теги показывают, что люди выбирают и как об этом говорят. Оценки — это уже более весомая реакция: как только что рассказанная история была воспринята. Здесь интересно смотреть на закономерности поведения аудитории.

### ❓ Вопрос 9: Как менялась активность выставления оценок с годами и как распределились баллы?

```python
dvis.distribution_by_key(
    ratings.movies.dist_by_year(),
    title="Ratings Activity by Year",
    max_bar_width=40,
)

%timeit ratings.movies.dist_by_year()
```

> Пик активности пришёлся на 2000 год (10 061 оценка) — но это, скорее, не бум кинопроизводства, а бум самого MovieLens: именно тогда сервис начал стремительно набирать пользователей, и они бросились переоценивать всё, что смотрели раньше.

```python
dvis.distribution_by_key(
    ratings.movies.dist_by_rating(),
    title="Rating Distribution",
    max_bar_width=40,
)

%timeit ratings.movies.dist_by_rating()
```

> Мы — оптимисты: любимая оценка зрителей — **4.0** (26,8 тыс. раз), а вовсе не «твёрдая тройка». Похоже, на MovieLens если фильм досмотрели до конца, ставят не «нормально», а «хорошо».

Число оценок — грубый, но честный индикатор культурного веса фильма: сколько людей вообще сочли нужным включить его и досмотреть до конца.

### ❓ Вопрос 10: Какие фильмы привлекли внимание рекордного числа зрителей?

```python
dvis.bar_chart(
    ratings.movies.top_by_num_of_ratings(5),
    title="Top 5 Movies by Number of Ratings",
    max_bar_width=30,
)

%timeit ratings.movies.top_by_num_of_ratings(5)
```

> Три титана 1994 года держат пьедестал: *Forrest Gump*, *The Shawshank Redemption* и *Pulp Fiction* — фильмы, которые проверяют время лучше любых блокбастеров своего десятилетия.

### ❓ Вопрос 11: Кто возглавляет рейтинг признания по среднему и медианному баллу?

```python
dvis.bar_chart(
    ratings.movies.top_by_ratings(5, metric="average"),
    title="Top 5 by Average Rating",
    max_bar_width=30,
)

%timeit ratings.movies.top_by_ratings(5, metric="average")
```

```python
dvis.bar_chart(
    ratings.movies.top_by_ratings(5, metric="median"),
    title="Top 5 by Median Rating",
    max_bar_width=30,
)

%timeit ratings.movies.top_by_ratings(5, metric="median")
```

> Осторожно: идеальные пятёрки почти всегда означают, что фильм оценили буквально пару раз — это не «лучший фильм всех времён», а математический шум малой выборки. Настоящих фаворитов будем искать иначе — в следующем, бонусном вопросе про «скрытые шедевры».

Согласие в оценках отражает общий вкус, но самое интересное происходит там, где мнения раскалываются, — это обычно граница между разными эстетическими или мировоззренческими лагерями внутри одной аудитории.

### ❓ Вопрос 12: Какие фильмы вызвали максимальную дисперсию мнений?

```python
dvis.bar_chart(
    ratings.movies.top_controversial(5),
    title="Most Controversial Movies (Variance)",
    max_bar_width=30,
)

%timeit ratings.movies.top_controversial(5)
```

> Самый глубокий раскол мнений вызывают не блокбастеры, а авторское кино — *«Иваново детство»* Тарковского и *«Фанни и Александр»* Бергмана. Зрители либо признают в них шедевр, либо категорически не принимают язык медленного, философского кино — равнодушных почти не остаётся.

Топ по количеству оценок — это про охват. Отдельный вопрос — какие фильмы не привлекли толпы, но заслужили стабильно высокую оценку тех немногих, кто их всё-таки нашёл.

### 💡 Бонусный вопрос 13: Где скрываются «неизвестные шедевры» с высочайшим баллом, но умеренным числом оценок?

```python
dvis.bar_chart(
    ratings.movies.hidden_gems(5, min_ratings=15, max_ratings=50),
    title="Hidden Gems (High score, moderate count)",
    max_bar_width=30,
)

%timeit ratings.movies.hidden_gems(5, min_ratings=15, max_ratings=50)
```

> Вот они, настоящие скрытые шедевры: *A Streetcar Named Desire* и *Sunset Blvd.* — фильмы полувековой давности, которые никогда не попадут в топ по числу оценок, но у тех немногих, кто их посмотрел, стабильно выбивают 4.3+ балла. Возможно, лучший фильм на вечер — вовсе не тот, что все сейчас обсуждают.

От фильмов — к самой аудитории. Кто эти люди, что скрываются за строчками датасета?

### ❓ Вопрос 14: А как ведут себя сами зрители? Кто ставит оценки чаще всего и кто из них наиболее непредсказуем?

```python
dvis.bar_chart(
    ratings.users.dist_by_num_of_ratings(),
    title="Users by Number of Ratings Given",
    max_bar_width=30,
)

%timeit ratings.users.dist_by_num_of_ratings()
```

```python
dvis.bar_chart(
    ratings.users.dist_by_rating(metric="average"),
    title="Users by Their Average Given Rating",
    max_bar_width=30,
)

%timeit ratings.users.dist_by_rating(metric="average")
```

```python
dvis.bar_chart(
    ratings.users.top_controversial(5),
    title="Most Unpredictable Users (Variance)",
    max_bar_width=30,
)

%timeit ratings.users.top_controversial(5)
```

### 💡 Бонусный вопрос 15: Кто доминирует в киносообществе — строгие критики или благодарные фанаты?

```python
dvis.bar_chart(
    ratings.users.critics_and_fans_ratio(),
    title="Audience Segmentation",
    max_bar_width=30,
)

%timeit ratings.users.critics_and_fans_ratio()
```

Здесь разворачивается главное несоответствие индустрии: одни фильмы сжигают бюджет и проваливаются, другие на скромные деньги делают кассу мечты.

Бюджет и прибыль — абсолютные цифры. Отдельный вопрос — эффективность каждого вложенного доллара.

# Итоги

Кино — молодое искусство, но оно унаследовало главную функцию рассказа историй: помогать людям понимать себя и друг друга. Разница в том, что теперь у этого процесса есть побочный продукт — данные, по которым можно проследить, чего хочет массовый зритель в конкретный момент времени.

1. **Динамика релизов и жанры**
   * Пик кинопроизводства в датасете пришелся на 2000–2009 годы (до 311 релизов в год).
   * Базовые жанры каталога — Drama (4361) и Comedy (3756): два самых универсальных эмоциональных запроса, радость и рефлексия. Sci-Fi и Fantasy стабильно входят в топ-10 на протяжении всей истории каталога — воображение будущего оказывается такой же постоянной потребностью, как и смех.

2. **Пользовательские теги**
   * Самый частый тег — служебный `In Netflix queue` (131 вызов). Среди содержательных лидируют `atmospheric`, `thought-provoking` и `surreal`.
   * Теги про «robots» и андроидов подтверждают устойчивый интерес к историям о будущем, уже увиденный в жанровой статистике.

3. **Оценки и поведение аудитории**
   * Распределение смещено в плюс: чаще всего ставят 4.0 (26.8k) и 3.0 (20k) — аудитория скорее снисходительна, чем строга (лишь 8% зрителей — стабильно жёсткие критики).
   * Наибольший раскол мнений вызывают не блокбастеры, а авторское кино (*«Иваново детство»*, *«Фанни и Александр»*) — там, где сталкиваются разные представления о том, каким должно быть кино.
   * Фильмы 40–50-х годов (*Sunset Blvd.*, *A Streetcar Named Desire*) стабильно удерживают средний балл ≥ 4.3 при небольшом числе голосов — признание есть, просто не массовое.

4. **Экономика и окупаемость (TMDb)**
   * Большой бюджет не гарантирует успеха: *Cutthroat Island* ($102M бюджета) осталась убыточной.
   * Рекордную окупаемость показывают не самые дорогие, а самые оригинальные проекты: *Toy Story* достигла ROI ≈ 32× ($932M чистой прибыли при бюджете $30M).

**Жанры, оценки и деньги** — это четыре разных способа зафиксировать один и тот же импульс: что людям на самом деле нужно от историй.
